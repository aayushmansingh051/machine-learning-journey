```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
import pandas as pd
import numpy as np
%matplotlib inline
```

```python
X,y=make_blobs(n_samples=1000,centers= 3,n_features=2)
```

```python
X
```

```python
y
```

```python
plt.scatter(X[:,0],X[:,1],c=y)
```

```python
#Standardization--feature scaling techinique 
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.5,random_state=42)
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

#.train is used to prevent form data leakage
```

```python
from sklearn.cluster import KMeans
```

# Elbow method to slect k values
The Elbow Method is a simple visual technique used in K‑Means clustering to decide the best number of clusters (k).

```python
wcss=[]
for k in range (1,11):
    Kmeans=KMeans(n_clusters=k,init='k-means++')
    Kmeans.fit(X_train_scaled)
    wcss.append(Kmeans.inertia_)
```

# plot the elbow curve

```python
plt.plot(range(1,11),wcss)
plt.xticks(range(1,11))
plt.xlabel("No. of clusters")
plt.ylabel("WCSS")
```

```python
Kmeans=KMeans(n_clusters=3,init='k-means++')
```

```python

from sklearn.cluster import KMeans
Kmeans.fit(X_train_scaled)
y_pred=Kmeans.predict(X_test_scaled)
y_pred
```

```python
plt.scatter(X_test[:,0],X_test[:,1],c=y_pred)
```

## Create KneeLocator object

```python
from kneed import KneeLocator

# Example WCSS values (replace with your own)



kl = KneeLocator(range(1, 11), wcss, curve='convex', direction='decreasing')

# Print the elbow point
print("Elbow point (best k):", kl.elbow)
kl.elbow
```

# shilhouette Score

```python
from sklearn.metrics import silhouette_score
for k in range(2,11):
    Kmeans=KMeans(n_clusters=k,init='k-means++')
    Kmeans.fit(X_train_scaled)
```

```python
score = silhouette_score(X_train_scaled, Kmeans.labels_)
```

```python
shilouette_coefficients=[]
shilouette_coefficients.append(score)
```

```python
shilouette_coefficients
```

```python
plt.plot(range (2,11),shilouette_coefficients)
plt.xlabel('No.of clusters')
plt.ylabel('shilouette_coefficients')
plt.title('Silhouette Score vs Number of Clusters')
plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

silhouette_coefficients = []

# Loop over k values
for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_train_scaled)
    score = silhouette_score(X_train_scaled, kmeans.labels_)
    silhouette_coefficients.append(score)

# Plot silhouette scores
plt.plot(range(2, 11), silhouette_coefficients)
plt.xlabel('Number of clusters')
plt.ylabel('Silhouette Coefficient')
plt.title('Silhouette Score vs Number of Clusters')
plt.show()
```
